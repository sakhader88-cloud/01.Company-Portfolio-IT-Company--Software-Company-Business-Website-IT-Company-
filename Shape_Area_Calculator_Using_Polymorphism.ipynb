```python
import math

class Shape:
    """Base class for all shapes."""
    def area(self):
        # We raise an error if the child class forgot to define this.
        raise NotImplementedError("Subclasses must implement the area method")

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * (self.radius ** 2)

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height
```

```python
import math
from abc import ABC, abstractmethod

# 1. The Abstract Base Class (The Blueprint)
class Shape(ABC):
    """
    Shape is an abstract base class. You cannot create a 'Shape' object directly.
    It serves as a contract: any class inheriting from Shape MUST define area().
    """

    @abstractmethod
    def area(self):
        """Calculate the area of the shape."""
        pass

    def __str__(self):
        """A helper method to identify the class name automatically."""
        return self.__class__.__name__

# 2. Derived Class: Circle
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        # Formula: π × r²
        return math.pi * (self.radius ** 2)

# 3. Derived Class: Rectangle
class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        # Formula: length × width
        return self.length * self.width

# 4. Derived Class: Triangle
class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        # Formula: ½ × base × height
        return 0.5 * self.base * self.height

# --- DEMONSTRATION OF POLYMORPHISM ---

def print_shape_details(shape_list):
    """
    This function demonstrates polymorphism. It treats every object
    in the list as a 'Shape', regardless of whether it's a Circle or Triangle.
    """
    print(f"{'Shape Type':<15} | {'Calculated Area':<15}")
    print("-" * 35)

    for shape in shape_list:
        # Polymorphism happens here: shape.area() behaves differently
        # based on the specific instance being processed.
        result = shape.area()
        print(f"{str(shape):<15} | {result:>15.2f}")

# Main execution
if __name__ == "__main__":
    # Creating a heterogeneous list (a mix of different shapes)
    my_shapes = [
        Circle(radius=5),
        Rectangle(length=10, width=4),
        Triangle(base=6, height=8),
        Circle(radius=2.5)
    ]

    print("Executing Polymorphic Area Calculations:\n")
    print_shape_details(my_shapes)
```
