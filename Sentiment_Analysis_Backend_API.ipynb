**Steps & Code**



I’ll build a Flask backend that:

Accepts text via POST request (like a mini AI API).

Analyzes sentiment (positive/negative/neutral) using TextBlob.

Optionally accepts file uploads (text files).

Runs live on Google Colab via ngrok.

Step 2 – Import and initialize

```python
from flask import Flask, request, jsonify
from flask_ngrok import run_with_ngrok
from textblob import TextBlob

app = Flask(__name__)
run_with_ngrok(app)  # starts ngrok tunnel automatically
```

Step 3 – Create sentiment route

```python
@app.route('/analyze', methods=['POST'])
def analyze_sentiment():
    # Check if text was sent directly
    if 'text' in request.form:
        text = request.form['text']
    # Or if a file was uploaded
    elif 'file' in request.files:
        file = request.files['file']
        text = file.read().decode('utf-8')
    else:
        return jsonify({'error': 'No text or file provided'}), 400

    # Analyze using TextBlob
    blob = TextBlob(text)
    sentiment = blob.sentiment.polarity

    if sentiment > 0:
        result = 'positive'
    elif sentiment < 0:
        result = 'negative'
    else:
        result = 'neutral'

    return jsonify({
        'text': text,
        'sentiment_score': sentiment,
        'sentiment_label': result
    })


@app.route('/')
def home():
    return "✅ Sentiment API is running! Use POST /analyze"
```

Step 4 – Run server

```python
@app.route('/')
def home():
    return "✅ Sentiment API is running! Use POST /analyze"

app.run()
```
