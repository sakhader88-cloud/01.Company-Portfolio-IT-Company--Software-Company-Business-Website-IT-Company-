# 🎓 Student Marks Analyzer

# This program asks for a student's marks, finds the average,

# highest and lowest score, and gives a performance comment.

# Step 1: Input section

```python
print("=== Student Marks Analyzer ===")
```

```python
name = input("Enter student name: ")
```


# We'll take marks for 5 subjects

```python
subjects = ["Math", "Science", "English", "History", "Computer"]
```

```python
marks = []
```

```python
for subject in subjects:
    mark = float(input(f"Enter marks for {subject}: "))
    marks.append(mark)
```

# Step 2: Processing / Calculations

```python
total = sum(marks)
average = total / len(marks)
highest = max(marks)
lowest = min(marks)
```

# Step 3: Decide grade or performance comment

```python
if average >= 90:
    comment = "Excellent! Keep it up 💪"
elif average >= 75:
    comment = "Very Good! You’re doing great 👍"
elif average >= 60:
    comment = "Good, but you can improve 👏"
elif average >= 40:
    comment = "Needs improvement 📘"
else:
    comment = "Failed 😔 — study harder next time."
```


# Step 4: Output results

```python
print("\n=== Report Card ===")
print(f"Student Name  : {name}")
for i in range(len(subjects)):
    print(f"{subjects[i]:<10}: {marks[i]}")
```

```python

print(f"\nTotal Marks   : {total}")
print(f"Average Marks : {average:.2f}")
print(f"Highest Marks : {highest}")
print(f"Lowest Marks  : {lowest}")
print(f"Performance   : {comment}")
```

💡 How it works

Takes user input for marks in 5 subjects.

Calculates total, average, highest, and lowest marks.

Evaluates performance using conditional logic.

Prints a formatted report card.
