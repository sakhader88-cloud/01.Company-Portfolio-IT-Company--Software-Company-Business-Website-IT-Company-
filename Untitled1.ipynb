```python
from abc import ABC, abstractmethod

# 1. Base Class (Abstract)
class Employee(ABC):
    def __init__(self, employee_id, name, department, base_salary):
        self.employee_id = employee_id
        self.name = name
        self.department = department
        self.base_salary = base_salary

    @abstractmethod
    def calculate_salary(self):
        """Must be implemented by all derived classes"""
        pass

    def display_details(self):
        print(f"[{self.__class__.__name__}] ID: {self.employee_id} | Name: {self.name} | Dept: {self.department}")

# 2. Derived Classes
class FullTimeEmployee(Employee):
    def __init__(self, employee_id, name, department, base_salary, bonus):
        super().__init__(employee_id, name, department, base_salary)
        self.bonus = bonus

    def calculate_salary(self):
        return self.base_salary + self.bonus

class PartTimeEmployee(Employee):
    def __init__(self, employee_id, name, department, hours_worked, hourly_rate):
        # Base salary isn't strictly used for calculation here, but we pass it as 0
        super().__init__(employee_id, name, department, 0)
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate

    def calculate_salary(self):
        return self.hours_worked * self.hourly_rate

class ContractEmployee(Employee):
    def __init__(self, employee_id, name, department, contract_amount):
        super().__init__(employee_id, name, department, contract_amount)
        self.contract_amount = contract_amount

    def calculate_salary(self):
        return self.contract_amount

# 3. Management System (Logic Layer)
class PayrollSystem:
    def __init__(self):
        self.employees = []

    def add_employee(self, emp):
        self.employees.append(emp)

    def show_all(self):
        print("\n--- All Employees ---")
        for emp in self.employees:
            emp.display_details()
            print(f"   Calculated Salary: ${emp.calculate_salary():,.2f}")

    def get_total_payroll(self):
        total = sum(emp.calculate_salary() for emp in self.employees)
        print(f"\nTotal Company Payroll: ${total:,.2f}")

    def filter_by_type(self, emp_type):
        print(f"\n--- Filtering by: {emp_type.__name__} ---")
        filtered = [e for e in self.employees if isinstance(e, emp_type)]
        for e in filtered:
            e.display_details()

# --- Execution ---
if __name__ == "__main__":
    sys = PayrollSystem()

    # Adding different types of employees
    sys.add_employee(FullTimeEmployee(101, "Alice Smith", "IT", 5000, 1200))
    sys.add_employee(PartTimeEmployee(102, "Bob Jones", "HR", 80, 25))
    sys.add_employee(ContractEmployee(103, "Charlie Brown", "Design", 4000))

    # Demonstrate Requirements
    sys.show_all()
    sys.get_total_payroll()
    sys.filter_by_type(FullTimeEmployee)
```
