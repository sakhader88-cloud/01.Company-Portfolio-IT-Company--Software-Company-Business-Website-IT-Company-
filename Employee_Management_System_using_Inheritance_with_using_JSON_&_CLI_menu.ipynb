```python
# --- Models (Inheritance & Polymorphism) ---

from abc import ABC, abstractmethod

class Employee(ABC):
  def __init__(self, employee_id, name, department, base_salary):
    self.employee_id = employee_id
    self.name = name
    self.department = department
    self.base_salary = base_salary

  @abstractmethod
  def calculate_salary(self):
    pass

  def to_dict(self):
    data = self.__dict__.copy()
    data['type'] = self.__class__.__name__
    return data

  def display_details(self):
    print(f"ID: {self.employee_id:<5} | Name: {self.name:<15} | Dept: {self.department:<12} | Type: {self.__class__.__name__}")
```

```python
class FullTimeEmployee(Employee):
  def __init__(self, employee_id, name, department, base_salary, bonus):
    super().__init__(employee_id, name, department, base_salary)
    self.bonus = bonus

  def calculate_salary(self):
    return self.base_salary + self.bonus
```

```python
def calculate_salary(self):
  return self.base_salary + self.bonus
```

```python
class PartTimeEmployee(Employee):
  def __init__(self, employee_id, name,department, hours_worked, hourly_rate):
    super().__init__(employee_id, name, department, 0)
    self.hours_worked = hours_worked
    self.hourly_rate = hourly_rate

  def calculate_salary(self):
    return self.hours_worked * self.hourly_rate
```

```python
def calculate_salary(self):
  return self.hours_worked * self.hourly_rate
```

```python
class ContractEmployee(Employee):
  def __init__(self, employee_id, name, department, contract_amount):
    super().__init__(employee_id, name, department, contract_amount)
    self.contract_amount = contract_amount

  def calculate_salary(self):
    return self.contract_amount
```

```python
def calculate_salary(self):
  return self.contract_amount
```

```python
# --- Data Layer ---
class DataHandler:
  FILE_NAME = "employees.json"

  @staticmethod
  def save(employees):
    with open(DataHandler.FILE_NAME, 'w') as f:
     json.dump([emp.to_dict() for emp in employees], f, indent=4)

  @staticmethod
  def load():
      if not os.path.exists(DataHandler.FILE_NAME):
          return []
      try:
          with open(DataHandler.FILE_NAME, 'r') as f:
              data = json.load(f)
              loaded = []
              for item in data:
                  etype = item.pop('type')
                  if etype == "FullTimeEmployee": loaded.append(FullTimeEmployee(**item))
                  elif etype == "PartTimeEmployee": loaded.append(PartTimeEmployee(**item))
                  elif etype == "ContractEmployee": loaded.append(ContractEmployee(**item))
              return loaded
      except (json.JSONDecodeError, KeyError):
          return []
```

```python
@staticmethod
def load():
    if not os.path.exists(DataHandler.FILE_NAME):
        return []
    try:
        with open(DataHandler.FILE_NAME, 'r') as f:
            data = json.load(f)
            loaded = []
            for item in data:
                etype = item.pop('type')
                if etype == "FullTimeEmployee": loaded.append(FullTimeEmployee(**item))
                elif etype == "PartTimeEmployee": loaded.append(PartTimeEmployee(**item))
                elif etype == "ContractEmployee": loaded.append(ContractEmployee(**item))
            return loaded
    except (json.JSONDecodeError, KeyError):
        return []
```

```python
class PayrollSystem:
  def __init__(self):
    self.employees = DataHandler.load()

  def menu(self):
    while True:
      print("\n--- EMPLOYEE MANAGEMENT SYSTEM ---")
      print("1. Add Employee")
      print("2. Display All Employees")
      print("3. View Total Payroll")
      print("4. Filter by Type")
      print("5. Exit")

      choice = input("\nSelect an option: ")

      if choice == '1': self.add_employee_ui()
      elif choice == '2': self.show_all()
      elif choice == '3': self.show_total_payroll()
      elif choice == '4': self.filter_ui()
      elif choice == '5': break
      else: print(" Invalid choice, try again.")

  def add_employee_ui(self):
    try:
      print("\nTypes: 1. Full-Time | 2. Part-Time | 3. Contract")
      type_choice = input("Select Type: ")
      eid = int(input("Enter ID: "))
      name = input("Enter Name: ")
      dept = input("Enter Department: ")

      employee = None
      if type_choice == '1':
        bonus = float(input("Enter Bonus: "))
        employee = FullTimeEmployee(eid, name, dept, 0, bonus)
      elif type_choice == '2':
        hours_worked = float(input("Enter Hours Worked: "))
        hourly_rate = float(input("Enter Hourly Rate: "))
        employee = PartTimeEmployee(eid, name, dept, hours_worked, hourly_rate)
      elif type_choice == '3':
        contract_amount = float(input("Enter Contract Amount: "))
        employee = ContractEmployee(eid, name, dept, contract_amount)
      else:
        print("Invalid employee type.")
        return

      self.employees.append(employee)
      DataHandler.save(self.employees)
      print(f"{employee.__class__.__name__} {name} added.")

    except ValueError:
      print("Invalid input. Please ensure numeric fields are entered correctly.")
    except Exception as e:
      print(f"An unexpected error occurred: {e}")

  def show_all(self):
    print("\n" + "="*60)
    for emp in self.employees:
      emp.display_details()
      print(f" Calculated Pay: ${emp.calculate_salary():,.2f}")
    print("="*60)

  def show_total_payroll(self):
    total = sum(emp.calculate_salary() for emp in self.employees)
    print(f"\n Total Company Payroll: ${total:, .2f}")

  def filter_ui(self):
    print("\n1. Full-Time | 2. Part-Time | 3. Contract")
    choice = input("Select type to filter: ")
    mapping = {'1': FullTimeEmployee, '2': PartTimeEmployee, '3': ContractEmployee}
    target = mapping.get(choice)

    if target:
      filtered = [e for e in self.employees if isinstance(e, target)]
      for e in filtered: e.display_details()
    else:
      print("Invalid Filter.")
```
